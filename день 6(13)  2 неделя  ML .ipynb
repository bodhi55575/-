```python
import numpy as np
import time

#  ЧАСТЬ 1: расчет ИМТ и фильтрация 

heights_m = np.array([1.60, 1.75, 1.82, 1.90, 1.70])
weights_kg = np.array([55, 85, 90, 110, 65])

# Считаем ИМТ сразу для всего массива
bmi_score = weights_kg / (heights_m ** 2)
print(f"Индекс массы тела: {np.round(bmi_score, 2)}")

# Фильтруем только тех, у кого ИМТ больше 25
high_risk_bmi = bmi_score[bmi_score > 25]
print(f"Значения выше 25: {np.round(high_risk_bmi, 2)}\n")


#  ЧАСТЬ 2: стандартизация данных 

def standardize(x):
    mean_val = np.mean(x)
    std_val = np.std(x)
    
    # Защита от деления на ноль при одинаковых значениях
    if std_val == 0:
        return np.full_like(x, 0.0, dtype=float)
        
    return (x - mean_val) / std_val

# Тестирование функции
data_normal = np.array([10.0, 20.0, 30.0, 40.0])
data_integers = np.array([5, 5, 5, 5]) 

result_normal = standardize(data_normal)
result_const = standardize(data_integers)

print(f"Нормализация обычных данных: {np.round(result_normal, 2)}")
print(f"Нормализация констант: {result_const}")
print(f"Тип данных при одинаковых значениях: {result_const.dtype}\n")


# ЧАСТЬ 3: сравнение скорости выполнения 

size = 1_000_000
test_list = list(range(size))
test_array = np.arange(size)

start_loop = time.time()
for _ in range(10):
    squares_list = [x ** 2 for x in test_list]
time_loop = time.time() - start_loop

start_np = time.time()
for _ in range(10):
    squares_array = test_array ** 2
time_np = time.time() - start_np

print(f"Cикл 'for':      {time_loop:.4f} сек.")
print(f"NumPy операция: {time_np:.4f} сек.")
print(f"\nУскорение: ~{int(time_loop/time_np)} раз(а)")
```
