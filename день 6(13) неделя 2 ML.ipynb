```python
import numpy as np

# ЧАСТЬ 1: Подготовка данных и генерация массивов

# Задаем данные: Рост (см) и Вес (кг)
raw_data = {
    'height': [160, 175, 182, 190, 170],  # Рост в сантиметрах
    'weight': [55,  85,  90,  110, 65]     # Вес в килограммах
}

# Превращаем словари в массивы float
heights_m = np.array(raw_data['height']) / 100.0  # переводим в метры
weights_kg = np.array(raw_data['weight'])

print("📊 Входные данные:")
print(f"Рост (м):  {heights_m}")
print(f"Вес (кг):  {weights_kg}\n")


# ЧАСТЬ 2: Векторизованные вычисления (Без цикла!)

# Формула: BMI = Weight / Height^2
# Оператор '** 2' возводит в квадрат ВСЕ числа в столбце сразу
bmi_score = weights_kg / (heights_m ** 2)

print(f"✅ Рассчитанный ИМТ: {np.round(bmi_score, 2)}\n")


# ЧАСТЬ 3: Фильтрация через Булеву Маску (Boolean Masking)


threshold = 25.0  # Пороговое значение

# 1. Создаем маску: условие возвращает True или False для каждого человека
is_at_risk = bmi_score > threshold 

print(f"🔍 Логика маски (ИМТ > {threshold}):")
print(is_at_risk) 

# 2. Применяем маску к исходному массиву с ИМТ
high_risk_people = bmi_score[is_at_risk]

print(f"\n⚠️ Люди в группе риска (ИМТ > {threshold}):")
print(high_risk_people)
```

```python

# ЧАСТЬ 2: Функция стандартизации (Z-score)

def standardize(x):
    """
    Нормализует массив по формуле Z = (x - mean) / std
    
    Параметры:
    x (np.array): Входной массив чисел.
    
    Возвращает:
    np.array: Массив со значениями Z-score.
              Если std=0, возвращает массив нулей (без ошибки).
    """
    
    # 1. Считаем статистику (среднее и разброс)
    mean_val = np.mean(x)
    std_val = np.std(x)
    
    # 2. Страховка от деления на ноль

    if std_val == 0:
        print(f"⚠️ Осторожно: Разброс равен 0 (std={std_val}).")
        print("   Деление невозможно. Возвращаю массив нулей той же длины.")
        return np.zeros_like(x) # Создаёт массив нулей такой же формы, как x
        
    # 3. Применяем формулу (Векторная операция!)
    result = (x - mean_val) / std_val
    
    return result


# --- ТЕСТЫ ---

print("🧪 Тест 1: Обычные данные")
data_normal = np.array([10.0, 20.0, 30.0, 40.0])
norm_data = standardize(data_normal)
print(f"Было: {data_normal}")
print(f"Стало (Z-score): {np.round(norm_data, 3)}")
# Среднее должно быть близко к 0, разброс к 1

print("\n🧪 Тест 2: Опасный случай (все числа одинаковые)")
data_constant = np.array([5.0, 5.0, 5.0, 5.0])
norm_const = standardize(data_constant)
print(f"Было: {data_constant}")
print(f"Стало (Z-score): {norm_const}")
# Тут сработала защита и вернула нули вместо ошибки!
```

```python
import numpy as np
import time # Импорт модуля времени для замеров

# Подготовим "тяжелые" данные (миллион чисел), 
# чтобы разница была видна глазу.

size = 1_000_000
data_list = list(range(size))       # Обычный список [0, 1, 2...]
data_array = np.arange(size)        # Быстрый массив NumPy [0, 1, 2...]


# ТЕСТ 1: Обычный цикл Python

start_loop = time.time() 

squares_loop = []
for x in data_list:
    squares_loop.append(x ** 2)     # Прибавляем квадрат каждого числа к новому списку

end_loop = time.time() 
time_taken_loop = end_loop - start_loop

print(f"⏳ Цикл 'for' занял: {time_taken_loop:.4f} сек.")



# ТЕСТ 2: Векторная операция NumPy
# одна строка кода обрабатывает весь массив целиком

start_np = time.time() 

squares_np = data_array ** 2         # Возводим все элементы сразу

end_np = time.time() 
time_taken_np = end_np - start_np

print(f"🚀 NumPy занял:      {time_taken_np:.4f} сек.")


# Итоговый вывод
print("\n🏆 Вывод:")
print(f"NumPy быстрее цикла примерно в {int(time_taken_loop / time_taken_np)} раз(а)!")
print("Это огромная разница при обучении реальных нейросетей.")
```
