```python

# КОНСОЛЬНЫЙ ПОМОЩНИК ПО ПОКУПКАМ (День 7: Финальная версия)
# Архитектура: ВВОД → ОБРАБОТКА → ВЫВОД


# --- КОНСТАНТЫ (настройки проекта) ---
DISCOUNT_THRESHOLD = 100.0  # Порог цены для скидки
DISCOUNT_RATE = 0.10        # Размер скидки (10%)


def input_positive_number(prompt: str) -> float:
    """Безопасный ввод положительного числа. Запрашивает повторно при ошибке."""
    while True:
        try:
            value = float(input(prompt))
            if value > 0:
                return value
            print("❌ Ошибка: значение должно быть больше нуля.")
        except ValueError:
            print("❌ Ошибка: введите корректное число (цифры и точка).")


def collect_purchases() -> list[float]:
    """ВВОД: запрашивает количество товаров и их цены. Возвращает список цен."""
    while True:
        try:
            count = int(input("Сколько всего товаров вы хотите добавить? "))
            if count >= 0:
                break
            print("❌ Ошибка: количество не может быть отрицательным.")
        except ValueError:
            print("❌ Ошибка: введите целое число.")

    cart = []
    for i in range(1, count + 1):
        price = input_positive_number(f"💰 Цена товара №{i}: ")
        cart.append(price)
    return cart


def calculate_statistics(cart: list[float]) -> dict:
    """ОБРАБОТКА: считает общую сумму, количество, средний чек и максимум."""
    if not cart:
        return {"sum": 0.0, "count": 0, "avg": 0.0, "max": 0.0}

    total = sum(cart)
    count = len(cart)
    return {
        "sum": total,
        "count": count,
        "avg": total / count,
        "max": max(cart)
    }


def apply_discount_and_calculate_total(cart: list[float]) -> dict:
    """ОБРАБОТКА: применяет скидку к товарам дороже порога. 
    Возвращает итоговую сумму и количество дорогих товаров."""
    discounted_total = 0.0
    expensive_count = 0

    for price in cart:
        if price > DISCOUNT_THRESHOLD:
            discounted_total += price * (1 - DISCOUNT_RATE)
            expensive_count += 1
        else:
            discounted_total += price

    return {"total_with_discount": discounted_total, "expensive_count": expensive_count}



# ГЛАВНЫЙ БЛОК: строго ВВОД → ОБРАБОТКА → ВЫВОД


print("🛒 Добро пожаловать в магазин!")
user_name = input("Как вас зовут? ").strip() or "Покупатель"
print(f"Здравствуйте, {user_name}!\n" + "="*30)

# 1️⃣ ВВОД ДАННЫХ
cart = collect_purchases()

# 2️⃣ ОБРАБОТКА ДАННЫХ
stats = calculate_statistics(cart)
pricing = apply_discount_and_calculate_total(cart)

# 3️⃣ ВЫВОД РЕЗУЛЬТАТОВ
print("\n СТАТИСТИКА ПОКУПОК:")
print(f"• Товаров: {stats['count']}")
print(f"• Сумма без скидок: {stats['sum']:.2f} ₽")
print(f"• Самая дорогая позиция: {stats['max']:.2f} ₽")
print(f"• Средний чек: {stats['avg']:.2f} ₽")

print("\n РАСЧЕТ СО СКИДКОЙ:")
print(f"• Товаров дороже {DISCOUNT_THRESHOLD} ₽: {pricing['expensive_count']}")
print(f"• ИТОГО К ОПЛАТЕ: {pricing['total_with_discount']:.2f} ₽")
print("="*30)
print("Спасибо за покупку! Возвращайтесь.")
```

### 1. В чём разница между `print()` и `return`?

`print()` только показывает текст на экране и ничего не передаёт дальше в программу. `return` отдаёт результат вычислений обратно в код, чтобы его можно было сохранить в переменную, передать в другую функцию или использовать в математике. Функции-калькуляторы обязаны использовать `return`, функции-оформители могут использовать `print()`.
