```python
# Функции по заданию:
# - safe_div(a, b): возвращает None при делении на ноль
# - is_palindrome(s): проверяет палиндром (игнорирует пробелы и регистр)

def safe_div(a, b):
    if b == 0:
        return None
    return a / b

def is_palindrome(s):
    cleaned = s.replace(" ", "").lower()
    return cleaned == cleaned[::-1]

# Проверка
print("safe_div(10, 0):", safe_div(10, 0))
# Исправлено по замечанию учителя: используем настоящий палиндром
print("is_palindrome('А роза упала на лапу Азора'):", is_palindrome("А роза упала на лапу Азора"))  # True
print("is_palindrome('сникерс'):", is_palindrome("сникерс"))  # False
```
