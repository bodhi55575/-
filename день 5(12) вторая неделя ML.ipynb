```python
# Задание 5: Анализ автомобильного текста (с детализацией)

# Импортируем функции
from utils import count_lines, get_unique_words, get_top_words

# Шаг 1: Проверяем файл
filename = 'text.txt'
print("=" * 50)
print(f"АНАЛИЗ ФАЙЛА: {filename}")
print("=" * 50)
print()

# Шаг 2: Считаем строки
lines = count_lines(filename)
print(f"1. Количество строк: {lines}")
print()

# Шаг 3: Анализируем уникальные слова
unique_before = set()
with open(filename, 'r', encoding='utf-8') as f:
    for line in f:
        for word in line.split():
            unique_before.add(word)
            
unique_after = get_unique_words(filename)
print(f"2. Уникальных слов:")
print(f"   - До очистки: {len(unique_before)}")
print(f"   - После очистки: {len(unique_after)}")
print(f"   - Пример первых 5: {list(unique_after)[:5]}")
print()

# Шаг 4: Анализируем топ-10 слов
top_words = get_top_words(filename, 10)
print("3. Топ-10 самых частых слов:")
for i, (word, count) in enumerate(top_words, 1):
    print(f"   {i}. {word}: {count} раз")
print()

# Шаг 5: Сохраняем результаты в CSV
with open('result.csv', 'w', encoding='utf-8') as f:
    f.write("Этап,Значение\n")
    f.write(f"Количество строк,{lines}\n")
    f.write(f"Уникальных слов до очистки,{len(unique_before)}\n")
    f.write(f"Уникальных слов после очистки,{len(unique_after)}\n")
    f.write("\nТоп-10 слов,Частота\n")
    for word, count in top_words:
        f.write(f"{word},{count}\n")
```
