```python
# Импорт библиотек для работы с путями и данными
from pathlib import Path
import utils # Импорт модуля, где хранятся функции

# ЧАСТЬ 1: Анализ входного файла (text.txt)

filename = 'text.txt'
source_path = Path(filename)

# Проверка наличия файла с помощью Path
if source_path.is_file():
    print(f"✅ Найден файл: {filename} ({source_path.stat().st_size} байт)")
else:
    raise FileNotFoundError(f"Файл '{filename}' отсутствует!")

# Выполняем анализ через наши функции из utils.py
line_count = utils.count_lines(filename)
print(f"\n1. Всего строк: {line_count}")

unique_words = utils.get_unique_words(filename)
print(f"2. Уникальных слов (после очистки): {len(unique_words)}")

top_10_words = utils.get_top_words(filename, k=10)
print("\n3. Топ-10 слов:")
for rank, (word, count) in enumerate(top_10_words, 1):
    print(f"   {rank}. {word}: {count} раз")



# ЧАСТЬ 2: Сохранение результатов (result.csv)

# Формируем контент для CSV вручную
csv_content = "Этап,Значение\n"
csv_content += f"Строк всего,{line_count}\n"
csv_content += f"Уникальных слов,{len(unique_words)}\n"
csv_content += "\nСлово,Частота\n"

for word, count in top_10_words:
    csv_content += f"{word},{count}\n"

# Запись файла одной командой с помощью Path
output_path = Path('result.csv')
output_path.write_text(csv_content, encoding='utf-8')

print(f"\n✅ Результат сохранён в файл: {output_path.name}")
print(f"   Размер готового файла: {output_path.stat().st_size} байт")
```
