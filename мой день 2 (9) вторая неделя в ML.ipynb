```python
# ЗАДАНИЕ: Мини-калькулятор
# Принимает 2 числа и оператор (+, -, *, /)
# Возвращает результат операции или None при ошибке

# Первое число
a = 10

# Оператор (+, -, *, /)
op = '+'

# Второе число
b = 5

if op == '+':
    # Сложение
    result = a + b
    
elif op == '-':
    # Вычитание
    result = a - b
    
elif op == '*':
    # Умножение
    result = a * b
    
elif op == '/' and b != 0:
    # Деление (только если делитель не равен 0!)
    result = a / b
    
else:
    # Если оператор неверный или деление на 0
    result = None

# Выводим результат
print(result)
```

```python
# ЗАДАНИЕ: Счётчик букв/цифр/пробелов
# Программа считает количество букв, цифр и пробелов в строке "porche 911 turbo"

text = "porche 911 turbo"  

letters = 0   # Счётчик для букв
digits = 0    # Счётчик для цифр
spaces = 0    # Счётчик для пробелов

# Цикл for перебирает КАЖДЫЙ символ в строке по одному
for char in text:
    # ПРОВЕРКА 1: Это буква?
    # .isalpha() возвращает True, если символ - буква (a-z, A-Z)
    if char.isalpha():
        # Увеличиваем счётчик букв на 1
        letters += 1
    
    # ПРОВЕРКА 2: Это цифра?
    # .isdigit() возвращает True, если символ - цифра (0-9)
    elif char.isdigit():
        # Увеличиваем счётчик цифр на 1
        digits += 1
    
    # ПРОВЕРКА 3: Это пробел?
    # Сравниваем символ с пробелом ' '
    elif char == ' ':
        # Увеличиваем счётчик пробелов на 1
        spaces += 1

# Используем f-строку для форматированного вывода результатов
print(f"Буквы: {letters}, Цифры: {digits}, Пробелы: {spaces}")
```

```python
# ЗАДАНИЕ: Поиск максимума и его индекса
# Без использования функции max()

numbers = [3, 7, 2, 9, 4]  

# Предполагаем, что первое число в списке - это максимум
max_value = numbers[0]  # Максимальное значение (начинаем с первого элемента)
max_index = 0           # Индекс максимального значения (начинаем с индекса 0)

# Начинаем с индекса 1, потому что индекс 0 уже проверили выше
for i in range(1, len(numbers)):
    # ПРОВЕРКА: Текущее число больше текущего максимума?
    if numbers[i] > max_value:
        # Если да, обновляем максимум и его индекс
        max_value = numbers[i]  # Запоминаем новое максимальное значение
        max_index = i           # Запоминаем индекс нового максимума

# Используем f-строку для форматированного вывода результатов
print(f"Максимум: {max_value}, Индекс: {max_index}")
```
