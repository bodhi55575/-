```python
# Задание 1: Нормализация через разные подходы

def normalize_loop(nums):
    """
    Нормализует список через обычный цикл.
    Возвращает список в диапазоне [0, 1].
    """
    min_val = min(nums)
    max_val = max(nums)
    
    if max_val == min_val:
        return [0.0] * len(nums)
    
    range_val = max_val - min_val
    result = []
    
    for num in nums:
        result.append((num - min_val) / range_val)
    
    return result


def normalize_comprehension(nums):
    """
    Нормализует список через списковое выражение.
    Возвращает список в диапазоне [0, 1].
    """
    min_val = min(nums)
    max_val = max(nums)
    
    if max_val == min_val:
        return [0.0] * len(nums)
    
    range_val = max_val - min_val
    return [(num - min_val) / range_val for num in nums]


# Проверка: оба метода дают одинаковый результат
test_data = [10, 20, 30, 40, 50]

print("Через цикл:", normalize_loop(test_data))
# [0.0, 0.25, 0.5, 0.75, 1.0]

print("Через comprehension:", normalize_comprehension(test_data))
# [0.0, 0.25, 0.5, 0.75, 1.0]
```

```python
# Задание 2: Пайплайн из функций (каждая решает одну задачу)

def filter_positive(numbers):
    """Фильтрует только положительные числа."""
    return [num for num in numbers if num > 0]


def square(numbers):
    """Возводит числа в квадрат."""
    return [num ** 2 for num in numbers]


def sum_numbers(numbers):
    """Суммирует числа."""
    return sum(numbers)


def process_pipeline(data, filter_func, transform_func, aggregate_func):
    """
    Общий пайплайн: фильтр → преобразование → агрегация.
    
    Параметры:
    - data: исходные данные
    - filter_func: функция фильтрации
    - transform_func: функция преобразования
    - aggregate_func: функция агрегации
    
    Возвращает:
    - результат обработки
    """
    filtered = filter_func(data)
    transformed = transform_func(filtered)
    return aggregate_func(transformed)


# Тестирование пайплайна
data = [5, -3, 2, -1, 8, 0, 4]

# Вариант 1: фильтрация → квадрат → сумма
result1 = process_pipeline(
    data,
    filter_positive,
    square,
    sum_numbers
)
print("Вариант 1 (фильр → квадрат → сумма):", result1)
# 25 + 4 + 64 + 16 = 109

# Вариант 2: фильтрация → квадрат → среднее
result2 = process_pipeline(
    data,
    filter_positive,
    square,
    lambda nums: sum(nums) / len(nums) if nums else 0
)
print("Вариант 2 (фильр → квадрат → среднее):", result2)
# (25 + 4 + 64 + 16) / 4 = 27.25
```

```python
# Задание 3: Безопасная нормализация с обработкой ошибок

def safe_normalize(nums):
    """
    Безопасная нормализация с обработкой ошибок.
    
    Возвращает:
    - нормализованный список
    - или сообщение об ошибке
    """
    try:
        min_val = min(nums)
        max_val = max(nums)
        
        if max_val == min_val:
            return [0.0] * len(nums)
        
        range_val = max_val - min_val
        return [(num - min_val) / range_val for num in nums]
    
    except TypeError:
        return "Ошибка: в списке есть нечисловые значения"
    except ValueError:
        return "Ошибка: список пустой"
    except Exception as e:
        return f"Неизвестная ошибка: {str(e)}"


# Тестирование
print(safe_normalize([10, 20, 30, 40, 50]))  # [0.0, 0.25, 0.5, 0.75, 1.0]
print(safe_normalize([5, 5, 5, 5]))          # [0.0, 0.0, 0.0, 0.0]
print(safe_normalize([]))                    # Ошибка: список пустой
print(safe_normalize([10, "20", 30]))        # Ошибка: в списке есть нечисловые значения
```
