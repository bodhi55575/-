```python
# 1 словарь частот + топ-3 самых частых слов
# Исходный текст
words = ['cat', 'dog', 'cat', 'bird', 'dog', 'cat', 'fish', 'dog', 'cat']

# Собираем словарь частот
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

print("Словарь частот:", freq)
# {'cat': 4, 'dog': 3, 'bird': 1, 'fish': 1}
```

```python
def top_k(scores: dict, k: int) -> list:
    """Возвращает список из k пар (ключ, значение) по убыванию частоты."""
    
    # Критерий сортировки:
    # -x[1] — ставит числа с большей частотой выше (за счет знака минус)
    # x[0] — сортирует слова по алфавиту, если частота одинаковая
    sorted_items = sorted(scores.items(), key=lambda x: (-x[1], x[0]))
    
    # Извлекаем первые k элементов в формате (ключ, значение)
    return [(word, freq) for word, freq in sorted_items[:k]]


# Пример работы:
data = {'cat': 4, 'dog': 3, 'bird': 1, 'fish': 1}
print(top_k(data, 3))
```

```python
# 3: Мини-демо — изменение "течёт" во вложенном списке

# Создаём исходный вложенный список
original = [[1, 2], [3, 4]]

# Создаём поверхностную копию
shallow = original.copy()

print("До изменения:")
print("original =", original)
print("shallow  =", shallow)

# Меняем элемент во вложенном списке
shallow[0][0] = 999

print("\nПосле изменения shallow[0][0] = 999:")
print("original =", original)  # ← изменился!
print("shallow  =", shallow)

print("\nПочему?")
print("original is shallow       :", original is shallow)        # False
print("original[0] is shallow[0]:", original[0] is shallow[0])   # True ← одна и та же ссылка!
```
