```python
# ### Задача 1. Сумма цифр числа (используем цикл while)

# Вводим число и сохраняем его копию для обработки
number = int(input("Введите целое число: ")) 
number = abs(number)
temp_number = number 

# Обнуляем переменную накопитель перед началом цикла
digit_sum = 0 

# Цикл идет ПОКА число содержит цифры (больше нуля)
while temp_number > 0:
    last_digit = temp_number % 10       # убираем последнюю цифру
    digit_sum += last_digit             # Прибавляем её к сумме
    temp_number //= 10                  # Удаляем последнюю цифру из числа

print(f"Сумма цифр числа {number} равна: {digit_sum}")
```

```python
# ### Задача 2. Подсчет четных чисел до N (используем счетчик)

n = int(input("До какого числа считаем (N)? "))

# Инициализируем переменную-счетчик нулем
count_even = 0  

# Проходим по всем числам от 1 до N включительно
for i in range(1, n + 1):
    
    # Если число четное (остаток от деления на 2 равен 0)
    if i % 2 == 0:
        # увеличиваем наш счетчик на единицу
        count_even += 1 

print(f"Найдено чётных чисел: {count_even}")
```

```python
# ### Задача №1.  Сумма цифр в отдельной функции

def sum_of_digits(n):
    """Считает сумму всех цифр переданного числа."""
    temp_n = n          # копия числа
    digit_sum = 0       # Накопитель
    
    while temp_n > 0:
        last_digit = temp_n % 10
        digit_sum += last_digit
        temp_n //= 10
        
    return digit_sum      # Возвращаем результат

# блок программы 
my_num = int(input("Введите большое число: "))
total = sum_of_digits(my_num)

print(f"Сумма цифр в числе {my_num} равна: {total}")
```

```python
# ### Задача №2. Площадь  с учетом скидки
# Функция принимает размеры и процент скидки, вычисляет итоговую стоимость.

def calc_floor_cost(length, width, price_per_sqm, discount_percent):
    """Вычисляет площадь и цену пола со скидкой."""
    
    # Считаем площадь
    area = length * width
    
    # Считаем цену без скидки
    total_price = area * price_per_sqm
    
    # Применяем скидку
    final_price = total_price - (total_price * (discount_percent / 100))
    
    return final_price

# --- Главное меню ---
l = float(input("Длина комнаты (м): "))
w = float(input("Ширина комнаты (м): "))
cost = float(input("Цена за м2: "))
disc = float(input("Скидка (%): "))

result = calc_floor_cost(l, w, cost, disc)
print(f"Итоговая цена закупки материалов: {result:.2f}")
```

```python
# ### Задача №3. Проверка текста на палиндром
# Функция возвращает логическое значение (True или False).

def is_palindrome(text):
    """Возвращает True, если текст читается одинаково влево и вправо."""
    # Приводим к нижнему регистру, чтобы "А" и "а" считались одинаковыми
    cleaned_text = text.lower().replace(" ", "")
    
    if cleaned_text == cleaned_text[::-1]:
        return True
    else:
        return False

# основной блок 
word = input("Напишите слово для проверки: ")

# Используем функцию в условии if напрямую
if is_palindrome(word):
    print(f"Слово '{word}' является палиндромом!")
else:
    print(f"Это просто слово.")
```

```python
# ### Задача №4. Подсчет гласных букв в тексте


def count_vowels(sentence):
    """Считает количество гласных букв в строке."""
    sentence = sentence.lower()   
    vowel_count = 0
    vowels = "aeiouAEIOUаеёиоуыэюя" # Список всех гласных
    
    for char in sentence:
        if char in vowels:
            vowel_count += 1
            
    return vowel_count

# Главный блок

phrase = input("Введите ваш текст: ")

vowels_total = count_vowels(phrase)
print(f"В вашем тексте найдено {vowels_total} гласных букв")
```

```python
# ### Задача №5. Генератор пароля 

def capitalize_words(text):
    """Превращает каждое слово в предложении с большой буквы."""
    # Метод .split() разбивает фразу на список слов
    words_list = text.split() 
    
    # Создаем новый список, где каждое слово идет с первой буквой в верхнем регистре (capitalized)
    new_words = []
    for word in words_list:
        # .capitalize() ставит 1й символ верхним регистром
        new_words.append(word.capitalize())
        
    # Собираем обратно в одну строку с пробелами
    result = " ".join(new_words)
    
    return result

# --- Главное меню ---
user_sentence = input("Введите предложение строчными буквами: ")

formatted_string = capitalize_words(user_sentence)
print(f"Готово: {formatted_string}")
```

```python
#
def get_positive_float(question_text):
    """Защищает программу от букв и отрицательных чисел. 
    Возвращает число (float)."""
    while True:
        try:
            # Пробуем превратить текст в число
            value = float(input(question_text))
            
            # Проверяем логику: цена должна быть больше нуля
            if value > 0:
                return value  # Успех! Возвращаем число наружу
            
            print("Ошибка! Цена не может быть меньше или равна нулю.")
            
        except ValueError:
            # СЮДА попадет программа, если введем БУКВЫ
            print("Ошибка ввода! Пожалуйста, вводите ТОЛЬКО цифры.")


def collect_purchases():
    """Собирает список покупок по запросу пользователя."""
    
    #  Проверка количества товаров 
    # используем защиту, чтобы пользователь случайно не закрыл программу буквами
    count = 0
    while True:
        try:
            s = input("Сколько всего покупок? ")
            count = int(s) # Пробуем превратить в целое число
            if count >= 0:
                break # Если всё ок, выходим из цикла запроса
            print("Нельзя вводить отрицательное количество!")
        except ValueError:
            print("Ошибка! Количество тоже должно быть числом.")

    cart = []
    
    # Цикл запрашивает цены столько раз, сколько указали
    for i in range(1, count + 1):
        # Вызываем функцию-защитник для каждой цены
        price = get_positive_float(f"Введите стоимость {i}-го товара: ")
        cart.append(price)
        
    return cart


def analyze_cart(cart):
    """Функция №3. Анализирует корзину."""
    
    if not cart: # Если корзина пустая
        return {"sum": 0, "count": 0, "avg": 0, "max": 0}

    total_sum = sum(cart)
    count_items = len(cart)
    average_check = total_sum / count_items
    most_expensive = max(cart)
    
    return {
        "total_sum": total_sum,
        "count_items": count_items,
        "average_check": average_check,
        "most_expensive": most_expensive
    }


def calculate_final_bill(cart):
    """Функция №4. Считает итоговую сумму со скидкой."""
    
    THRESHOLD = 100  # Порог цены
    DISCOUNT_RATE = 0.10 
    
    final_amount = 0
    
    for price in cart:
        if price > THRESHOLD:
            # Товар дороже порога — берем скидку
            final_amount += price * (1 - DISCOUNT_RATE)
        else:
            # Иначе платим полную цену
            final_amount += price
            
    expensive_count = sum(1 for p in cart if p > THRESHOLD)
    
    return {
        "bill": final_amount,
        "expensive_items_count": expensive_count
    }


# основной блок программы

name = input("Как вас зовут? ")
print(f"Добро пожаловать в магазин, {name}!")
print("-" * 30)

# Запускаем сбор покупок
items = collect_purchases()

if items:
    stats = analyze_cart(items)
    
    print("\n--- СТАТИСТИКА ---")
    print(f"Количество товаров: {stats['count_items']}")
    print(f"Общая сумма (без скидок): {stats['total_sum']:.2f}")
    print(f"Самая дорогая покупка: {stats['most_expensive']:.2f}")
    print(f"Средний чек: {stats['average_check']:.2f}")
    
    final_data = calculate_final_bill(items)
    
    print("\n--- ФИНАЛЬНЫЙ РАСЧЕТ ---")
    print(f"Дорогих товаров (>=100р): {final_data['expensive_items_count']}")
    print(f"ИТОГО К ОПЛАТЕ СО СКИДКОЙ: {final_data['bill']:.2f}")
else:
    print("Вы ничего не купили. Программа завершена.")
```

```python
# КОНСОЛЬНЫЙ ПОМОЩНИК ПО ПОКУПКАМ (День 7: Финальная версия)
# Архитектура: ВВОД → ОБРАБОТКА → ВЫВОД


# --- КОНСТАНТЫ (настройки проекта) ---
DISCOUNT_THRESHOLD = 100.0  # Порог цены для скидки
DISCOUNT_RATE = 0.10        # Размер скидки (10%)
import math

def input_positive_number(prompt: str) -> float:
    """Безопасный ввод положительного числа. Запрашивает повторно при ошибке."""
    while True:
        try:
            value = float(input(prompt))
            if value > 0 and math.isfinite(value):
                return value
            print("❌ Ошибка: значение должно быть больше нуля.")
        except ValueError:
            print("❌ Ошибка: введите корректное число (цифры и точка).")


def collect_purchases() -> list[float]:
    """ВВОД: запрашивает количество товаров и их цены. Возвращает список цен."""
    while True:
        try:
            count = int(input("Сколько всего товаров вы хотите добавить? "))
            if count >= 0:
                break
            print("❌ Ошибка: количество не может быть отрицательным.")
        except ValueError:
            print("❌ Ошибка: введите целое число.")

    cart = []
    for i in range(1, count + 1):
        price = input_positive_number(f"💰 Цена товара №{i}: ")
        cart.append(price)
    return cart


def calculate_statistics(cart: list[float]) -> dict:
    """ОБРАБОТКА: считает общую сумму, количество, средний чек и максимум."""
    if not cart:
        return {"sum": 0.0, "count": 0, "avg": 0.0, "max": 0.0}

    total = sum(cart)
    count = len(cart)
    return {
        "sum": total,
        "count": count,
        "avg": total / count,
        "max": max(cart)
    }


def apply_discount_and_calculate_total(cart: list[float]) -> dict:
    """ОБРАБОТКА: применяет скидку к товарам дороже порога. 
    Возвращает итоговую сумму и количество дорогих товаров."""
    discounted_total = 0.0
    expensive_count = 0

    for price in cart:
        if price > DISCOUNT_THRESHOLD:
            discounted_total += price * (1 - DISCOUNT_RATE)
            expensive_count += 1
        else:
            discounted_total += price

    return {"total_with_discount": discounted_total, "expensive_count": expensive_count}



# ГЛАВНЫЙ БЛОК: строго ВВОД → ОБРАБОТКА → ВЫВОД


print("🛒 Добро пожаловать в магазин!")
user_name = input("Как вас зовут? ").strip() or "Покупатель"
print(f"Здравствуйте, {user_name}!\n" + "="*30)

# 1️⃣ ВВОД ДАННЫХ
cart = collect_purchases()

# 2️⃣ ОБРАБОТКА ДАННЫХ
stats = calculate_statistics(cart)
pricing = apply_discount_and_calculate_total(cart)

# 3️⃣ ВЫВОД РЕЗУЛЬТАТОВ
print("\n СТАТИСТИКА ПОКУПОК:")
print(f"• Товаров: {stats['count']}")
print(f"• Сумма без скидок: {stats['sum']:.2f} ₽")
print(f"• Самая дорогая позиция: {stats['max']:.2f} ₽")
print(f"• Средний чек: {stats['avg']:.2f} ₽")

print("\n РАСЧЕТ СО СКИДКОЙ:")
print(f"• Товаров дороже {DISCOUNT_THRESHOLD} ₽: {pricing['expensive_count']}")
print(f"• ИТОГО К ОПЛАТЕ: {pricing['total_with_discount']:.2f} ₽")
print("="*30)
print("Спасибо за покупку! Возвращайтесь.")
```

### Ответы на контрольные вопросы:

**1. Что возвращает `input()`?**
Всегда возвращает строку (текст). Даже если ввести цифры, Python считает их текстом, поэтому для математики нужно использовать `int()` или `float()`.

**2. Когда нужен `if`, а когда цикл?**
`if` нужен для разового выбора пути (если А, то сделай Б). Цикл нужен, когда действие нужно повторить много раз (пройтись по всем товарам в корзине).

**3. Отличие `for` от `while`?**
`for` используется, когда мы заранее знаем количество повторений (например, цикл от 1 до 10). `while` используется, когда количество неизвестно, и цикл крутится, пока условие истинно (например, пока пароль неверный).

**4. Зачем правильно инициализировать счётчик/сумму?**
Переменные нужно обнулять (`sum = 0`) до цикла, иначе Python выдаст ошибку или прибавит новые данные к мусору из памяти. Для суммы база — 0, для умножения (факториал) — 1.

**5. Зачем нужен `return`?**
Чтобы функция отдала результат обратно в программу для дальнейших вычислений. Без `return` функция просто напечатает текст на экран, и мы не сможем использовать эти цифры в математике.

**6. Как тестировать программу?**
Проверять на граничных значениях: 0 товаров, 1 товар, отрицательные числа, ввод букв вместо цифр. Программа не должна падать с красной ошибкой, а должна просить повторить ввод.
