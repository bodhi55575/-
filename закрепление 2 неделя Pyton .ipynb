```python
# ### Задача 1. Сумма цифр числа (используем цикл while)

# Вводим число и сохраняем его копию для обработки
number = int(input("Введите целое число: ")) 
temp_number = number 

# Обнуляем переменную накопитель перед началом цикла
digit_sum = 0 

# Цикл идет ПОКА число содержит цифры (больше нуля)
while temp_number > 0:
    last_digit = temp_number % 10       # убираем последнюю цифру
    digit_sum += last_digit             # Прибавляем её к сумме
    temp_number //= 10                  # Удаляем последнюю цифру из числа

print(f"Сумма цифр числа {number} равна: {digit_sum}")
```

```python
# ### Задача 2. Подсчет четных чисел до N (используем счетчик)

n = int(input("До какого числа считаем (N)? "))

# Инициализируем переменную-счетчик нулем
count_even = 0  

# Проходим по всем числам от 1 до N включительно
for i in range(1, n + 1):
    
    # Если число четное (остаток от деления на 2 равен 0)
    if i % 2 == 0:
        # увеличиваем наш счетчик на единицу
        count_even += 1 

print(f"Найдено чётных чисел: {count_even}")
```

```python
# ### Задача 3. Проверка корректности ввода (повторный ввод до верного ответа)

age = -1 # Задаем явно неправильное значение, чтобы войти в цикл

# Пока возраст меньше нуля (условие истинно)
while age < 0:
    print("Ошибка! Возраст не может быть отрицательным.")
    
    try:
        # Спрашиваем ввод заново. Если число положительное — выходим из цикла
        age = int(input("Введите ваш реальный возраст (>0): "))
    except ValueError:
        print("Пожалуйста, введите только цифры!")

# Код ниже выполнится только когда условие цикла станет ложным
print(f"Ваш возраст подтвержден: {age}")
```

```python
# ### Задача 4. Вычисление факториала (цикл с накопителем умножения)

n = int(input("Вычислим факториал числа: "))

# Для умножения стартовое значение должно быть 1, иначе всё  0
factorial = 1 

for i in range(1, n + 1):
    # Умножаем текущее накопленное значение на новое число
    factorial *= i 
    
print(f"{n}! = {factorial}")
```

```python
# ### Задача №1.  Сумма цифр в отдельной функции

def sum_of_digits(n):
    """Считает сумму всех цифр переданного числа."""
    temp_n = n          # копия числа
    digit_sum = 0       # Накопитель
    
    while temp_n > 0:
        last_digit = temp_n % 10
        digit_sum += last_digit
        temp_n //= 10
        
    return digit_sum      # Возвращаем результат

# блок программы 
my_num = int(input("Введите большое число: "))
total = sum_of_digits(my_num)

print(f"Сумма цифр в числе {my_num} равна: {total}")
```

```python
# ### Задача №2. Площадь  с учетом скидки
# Функция принимает размеры и процент скидки, вычисляет итоговую стоимость.

def calc_floor_cost(length, width, price_per_sqm, discount_percent):
    """Вычисляет площадь и цену пола со скидкой."""
    
    # Считаем площадь
    area = length * width
    
    # Считаем цену без скидки
    total_price = area * price_per_sqm
    
    # Применяем скидку
    final_price = total_price - (total_price * (discount_percent / 100))
    
    return final_price

# --- Главное меню ---
l = float(input("Длина комнаты (м): "))
w = float(input("Ширина комнаты (м): "))
cost = float(input("Цена за м2: "))
disc = float(input("Скидка (%): "))

result = calc_floor_cost(l, w, cost, disc)
print(f"Итоговая цена закупки материалов: {result:.2f}")
```

```python
# ### Задача №3. Проверка текста на палиндром
# Функция возвращает логическое значение (True или False).

def is_palindrome(text):
    """Возвращает True, если текст читается одинаково влево и вправо."""
    # Приводим к нижнему регистру, чтобы "А" и "а" считались одинаковыми
    cleaned_text = text.lower().replace(" ", "")
    
    if cleaned_text == cleaned_text[::-1]:
        return True
    else:
        return False

# основной блок 
word = input("Напишите слово для проверки: ")

# Используем функцию в условии if напрямую
if is_palindrome(word):
    print(f"Слово '{word}' является палиндромом!")
else:
    print(f"Это просто слово.")
```

```python
# ### Задача №4. Подсчет гласных букв в тексте


def count_vowels(sentence):
    """Считает количество гласных букв в строке."""
    vowel_count = 0
    vowels = "aeiouAEIOUаеёиоуыэюя" # Список всех гласных
    
    for char in sentence:
        if char in vowels:
            vowel_count += 1
            
    return vowel_count

# Главный блок

phrase = input("Введите ваш текст: ")

vowels_total = count_vowels(phrase)
print(f"В вашем тексте найдено {vowels_total} гласных букв")
```

```python
# ### Задача №5. Генератор пароля 

def capitalize_words(text):
    """Превращает каждое слово в предложении с большой буквы."""
    # Метод .split() разбивает фразу на список слов
    words_list = text.split() 
    
    # Создаем новый список, где каждое слово идет с первой буквой в верхнем регистре (capitalized)
    new_words = []
    for word in words_list:
        # .capitalize() ставит 1й символ верхним регистром
        new_words.append(word.capitalize())
        
    # Собираем обратно в одну строку с пробелами
    result = " ".join(new_words)
    
    return result

# --- Главное меню ---
user_sentence = input("Введите предложение строчными буквами: ")

formatted_string = capitalize_words(user_sentence)
print(f"Готово: {formatted_string}")
```
